```
import torch
import torchvision
import torch.nn as nn
from torchvision import transforms

from PIL import Image
import pickle
import matplotlib.pyplot as plt
```

## Question 1
### Define Input Transformation

```
MEAN = [0.485, 0.456, 0.406]
STD = [0.229, 0.224, 0.225]
IMAGE_SIZE = 224
image_transforms = transforms.Compose([
                           transforms.Resize(IMAGE_SIZE),
                           transforms.CenterCrop(IMAGE_SIZE),
                           transforms.ToTensor(),
                           transforms.Normalize(mean = MEAN, std = STD)])

index2name = pickle.load(open("imagenet_class_names.pkl", "rb"))
```

### Define Model

```
model = torchvision.models.resnet50(pretrained=True)
softmax_layer = nn.Softmax(dim=1)
model.eval()
```

### Classify and Visualize Image

```
# define all images
image_paths = [ "dog.jpg", "cat.jpg", "car.jpg", "airplane.jpg", "banana.jpg", "bicycle.jpg", "bird.jpg", "keyboard.jpg", "boat.jpg", "truck.jpg" ]


# now get and show predictions
def show_predictions(model, model_name, image_paths, topk=5):
    softmax_layer = nn.Softmax(dim=1)
    model.eval()

    # go thru each image
    for image_path in image_paths:
        original_image = Image.open(image_path).convert("RGB")
        image = image_transforms(original_image).unsqueeze(0)

        with torch.no_grad():
            output = softmax_layer(model(image))

        top_probs, top_indices = torch.topk(output, topk, dim=1)

        plt.figure(figsize=(4, 4))
        plt.imshow(original_image)
        plt.title(f"{model_name}: {image_path}")
        plt.axis("off")
        plt.show()

        print(f"Top-{topk} predictions for {image_path}:")
        for i in range(topk):
            idx = top_indices[0, i].item()
            prob = top_probs[0, i].item() * 100
            print(f"{i + 1}. {index2name[idx]}: {prob:.2f}%")
        print("-" * 50)

resnet50 = torchvision.models.resnet50(pretrained=True)
resnet101 = torchvision.models.resnet101(pretrained=True)
densenet121 = torchvision.models.densenet121(pretrained=True)

show_predictions(resnet50, "ResNet-50", image_paths, topk=5)
show_predictions(resnet101, "ResNet-101", image_paths, topk=5)
show_predictions(densenet121, "DenseNet-121", image_paths, topk=5)
```

## Question 2

### Define Training Data Loader

```
train_batch_size = 32

train_transform = transforms.Compose([
    transforms.RandomHorizontalFlip(),
    transforms.RandomCrop(32, padding=4),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

train_dataset = torchvision.datasets.CIFAR10(
    root="./data",
    train=True,
    download=True,
    transform=train_transform
)

train_loader = torch.utils.data.DataLoader(
    train_dataset,
    batch_size=train_batch_size,
    shuffle=True,
    num_workers=2
)
```

### Define Model and Training Framework

```
class CIFAR10CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, 256),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(256, 10)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


cuda = torch.cuda.is_available()
model = CIFAR10CNN()

if cuda:
    model.cuda()

model.train()
learning_rate = 0.001
num_epochs = 3
optim = torch.optim.Adam(model.parameters(), lr=learning_rate)
loss_fn = nn.CrossEntropyLoss()
loss_logger = []
accuracy_logger = []

for epoch in range(1, num_epochs + 1):
    for it, (images, labels) in enumerate(train_loader):
        if cuda:
            images = images.cuda()
            labels = labels.cuda()

        outputs = model(images)
        loss = loss_fn(outputs, labels)

        optim.zero_grad()
        loss.backward()
        optim.step()

        predictions = torch.argmax(outputs, dim=1)
        accuracy = (predictions == labels).float().mean().item() * 100

        loss_logger.append(loss.item())
        accuracy_logger.append(accuracy)

        if it % 200 == 0:
            print("Epoch [%d/%d]. Iter [%d/%d]. Loss: %0.4f. Accuracy: %.2f" %
                  (epoch, num_epochs, it + 1, len(train_loader), loss.item(), accuracy))

torch.save(model.state_dict(), "CIFAR10-Model.pth")

plt.figure(figsize=(8, 5))
plt.plot(loss_logger)
plt.title("Training Loss over Iterations")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(accuracy_logger)
plt.title("Training Accuracy over Iterations")
plt.xlabel("Iteration")
plt.ylabel("Accuracy (%)")
plt.show()
```

### Load Model and Evaluate Model On Testing Dataset

```
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cuda = torch.cuda.is_available()
model = CIFAR10CNN()

if cuda:
    model.cuda()

model.load_state_dict(torch.load("CIFAR10-Model.pth"))
model.eval()

test_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

test_dataset = torchvision.datasets.CIFAR10(
    root="./data",
    train=False,
    download=True,
    transform=test_transform
)

test_loader = torch.utils.data.DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2
)

final_accuracy = 0.0
final_top_5_accuracy = 0.0

all_preds = []
all_labels = []
correct_predictions = []

class_names = ['Airplane', 'Automobile', 'Bird', 'Cat', 'Deer',
               'Dog', 'Frog', 'Horse', 'Ship', 'Truck']

with torch.no_grad():
    for it, (images, labels) in enumerate(test_loader):
        if cuda:
            images = images.cuda()
            labels = labels.cuda()

        outputs = model(images)

        preds = torch.argmax(outputs, dim=1)
        accuracy = (preds == labels).float().mean().item() * 100

        top5 = torch.topk(outputs, 5, dim=1).indices
        top_5_accuracy = top5.eq(labels.view(-1, 1)).any(dim=1).float().mean().item() * 100

        final_accuracy += accuracy
        final_top_5_accuracy += top_5_accuracy

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

        for i in range(len(images)):
            if preds[i] == labels[i] and len(correct_predictions) < 10:
                prob = torch.softmax(outputs[i], dim=0)[preds[i]].item() * 100
                correct_predictions.append((
                    images[i].cpu(),
                    labels[i].cpu().item(),
                    preds[i].cpu().item(),
                    prob
                ))

        if it % 500 == 0:
            print("Iter [%d/%d]. Accuracy: %.2f" % (it + 1, len(test_loader), accuracy))

print("Final Accuracy: %0.2f" % (final_accuracy / len(test_loader)))
print("Top 5 Accuracy: %0.2f" % (final_top_5_accuracy / len(test_loader)))

cm = confusion_matrix(all_labels, all_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)

fig, ax = plt.subplots(figsize=(10, 10))
disp.plot(ax=ax, xticks_rotation=45)
plt.title("Confusion Matrix")
plt.show()

def unnormalize(img):
    return img * 0.5 + 0.5

for img, true_label, pred_label, prob in correct_predictions:
    img = unnormalize(img)
    img = img.permute(1, 2, 0).numpy()

    plt.figure(figsize=(3, 3))
    plt.imshow(img)
    plt.title(f"Class: {class_names[pred_label]}. Probability: {prob:.2f}%")
    plt.axis("off")
    plt.show()
```
